толщина снежного покрова (см) - потом может сыграть, ибо есть колонки с мм -> нужно приводить к одной системе измерения. nan можно заменить на -1, Снежный покров не постоянный - 0.1

```python
df_mine['snow_depth_cm'].unique()
```

```python
# Заменяем значения в столбце 'snow_depth_cm' согласно условиям
df_mine['snow_depth_mm'] = df_mine['snow_depth_cm'].replace({'Менее 0.5': 0.5, 
                                                             'Снежный покров не постоянный.': 0.01, 
                                                             np.nan: -0.1})

# Переводим значения из строкового типа в тип с плавающей точкой (float)
df_mine['snow_depth_mm'] = df_mine['snow_depth_mm'].astype(float)

# Умножаем все значения в столбце 'snow_depth_cm' на 10
df_mine['snow_depth_mm'] = df_mine['snow_depth_mm'].mul(10)
df_mine['snow_depth_mm'].unique()
```

Уровень осадков в мм. непонятно, почему есть nan и осадков нет. nan - -1, осадков нет - 0, следы осадков - 0.01

```python
df_mine['precipitation_amount_mm'].unique()
```

```python
# Заменяем значения в столбце 'new_precipitation_amount_mm' согласно условиям
df_mine['new_precipitation_amount_mm'] = df_mine['precipitation_amount_mm'].replace({'Осадков нет': 0, 
                                                                                    'Следы осадков': 0.01, 
                                                                                    np.nan: -1})
df_mine['new_precipitation_amount_mm'].unique()
```

время накопления осадков в часах - или человеческим языком - интенсивность осадков, чем меньше число - тем сильнее осадки. nan замени на -1 симметрично с new_precipitation_amount_mm и для избегания деления на 0

```python
# Заменяем значения в столбце 'precipitation_accumulation_time_hrs' согласно условиям
df_mine['new_precipitation_accumulation_time_hrs'] = df_mine['precipitation_accumulation_time_hrs'].replace({np.nan: -1})
df_mine['new_precipitation_accumulation_time_hrs'].unique()
```

сложив два столбца, получилось, что значения nan симметричны. можно их заменить на 0. и затем посчитать плотность выпадения осадков

```python
df_mine_precipitation = pd.DataFrame({
    'precipitation': df_mine['precipitation_accumulation_time_hrs'].astype(str) + df_mine['precipitation_amount_mm'].astype(str)})
df_mine_precipitation['precipitation'].unique()
```

ввожу новый столбец - плотность выпадения осадков.

```python
# Устанавливаем формат вывода значений с плавающей точкой
pd.set_option('display.float_format', lambda x: '%.3f' % x)

df_mine['precipitation_density'] = df_mine['new_precipitation_amount_mm'].astype(float) / df_mine['new_precipitation_accumulation_time_hrs'].astype(float)
df_mine['precipitation_density'].round(3)
df_mine['precipitation_density'] = df_mine['precipitation_density'].replace({-np.inf: 0})
df_mine['precipitation_density'].unique()
```

температура воздуха, при которой начинается конденсация влаги из воздуха в виде росы. Это важный параметр для определения влажности воздуха и предсказания возможного образования осадков


Для ранжировки значений температуры точки росы, важно учитывать, что более высокие значения указывают на более высокую влажность воздуха. Чем ближе температура точки росы к текущей температуре воздуха, тем выше влажность воздуха. Ниже приведены общие интерпретации значений температуры точки росы и их связь с влажностью воздуха и вероятностью осадков:

- Температура точки росы ниже -10°C:
    - Очень сухой воздух.
    - Вероятность осадков низка.

- Температура точки росы от -10°C до 0°C:
    - Низкая влажность воздуха.
    - Вероятность осадков низкая до умеренной.

- Температура точки росы от 0°C до +10°C:
    - Умеренная влажность воздуха.
    - Вероятность осадков умеренная.

- Температура точки росы от +10°C до +15°C:
    - Высокая влажность воздуха.
    - Вероятность осадков высокая, особенно при наличии других факторов, таких как атмосферное движение и тепловой поток.

- Температура точки росы выше +15°C:
    - Очень высокая влажность воздуха.
    - Вероятность осадков очень высокая, особенно в условиях стабильного атмосферного движения и большого количества влаги в атмосфере.
